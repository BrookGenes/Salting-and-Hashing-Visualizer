```python
import hashlib
import secrets

def make_hash(password):
    return hashlib.sha256(password.encode()).hexdigest()

def salted_hash(password, salt):
    return hashlib.sha256((salt + password).encode()).hexdigest()

# Three made-up test passwords
test_passwords = ["BlueCat123!", "PizzaMoon45!", "TestPassword789!"]

for password in test_passwords:
    salt1 = secrets.token_hex(8)
    salt2 = secrets.token_hex(8)

    print("Test password:", password)
    print("Normal hash:", make_hash(password))
    print("Salted hash 1:", salted_hash(password, salt1))
    print("Salted hash 2:", salted_hash(password, salt2))
    print("Different salts create different hashes:",
          salted_hash(password, salt1) != salted_hash(password, salt2))
    print("-" * 50)
```
